#### finding the quantiles for each service and then converting them into a data frame for easy calculation

```python
quantiles = services_weekly.groupby('service')['patients_request'].quantile([0.25, 0.75])
quantiles = quantiles.unstack()
quantiles 
```

```python
quantiles['IQR'] = quantiles[0.75] - quantiles[0.25]
quantiles
```

```python
quantiles['Upper Bound']= quantiles[0.75] + (1.5 * quantiles[0.25])
quantiles
```

#### question 2 : Which services experienced unusually high patient demand, and during which weeks did those spikes occur?

```python
# Step 1: Add each service's upper bound to the weekly dataset
services_weekly['Upper Bound'] = services_weekly['service'].map(quantiles['Upper Bound'])

# Step 2: Flag unusually high-demand weeks
services_weekly['High Demand Outlier'] = (
    services_weekly['patients_request'] > services_weekly['Upper Bound']
)

# Step 3: Filter and display the outlier records
high_demand_weeks = services_weekly[
    services_weekly['High Demand Outlier']
]

# Step 4: Display relevant columns
print(
    high_demand_weeks[
        ['service', 'week', 'patients_request', 'Upper Bound']
    ].sort_values(by='patients_request', ascending=False)
)
```

#### Calculates the average of the four existing service-level refusal rates.

```python
weekly_refusals = (
    services_weekly.groupby('week')['Refusal_Rate']
    .mean()
    .sort_values(ascending=False)
)

print(weekly_refusals.head(5))
```

#### Calculates the overall hospital refusal rate using the total patients refused divided by total patients requested.

```python

weekly_refusals = services_weekly.groupby('week').agg({
    'patients_refused': 'sum',
    'patients_request': 'sum'
})

weekly_refusals['Refusal Rate %'] = (
    weekly_refusals['patients_refused'] /
    weekly_refusals['patients_request'] * 100
).round(2)

weekly_refusals = weekly_refusals.sort_values(
    by='Refusal Rate %',
    ascending=False
)

print(weekly_refusals.head(5))

```

#### What is the relationship between weekly patient demand and staff attendance?

```python
weekly_demand = (
    services_weekly.groupby('week')['patients_request']
    .sum()
    .reset_index()
)

weekly_attendance = (
    staff.groupby('week')['present']
    .mean()
    .mul(100)
    .reset_index(name='Attendance %')
)

merged_data1 = pd.merge(
    weekly_demand,
    weekly_attendance,
    on='week',
    how='inner'
)
correlation = merged_data1['patients_request'].corr(
    merged_data1['Attendance %']
)

print(correlation)
```
